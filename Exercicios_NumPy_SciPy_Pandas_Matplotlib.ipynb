### Exercício 19
Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

# Array 1 com 20 números inteiros aleatórios
array1 = np.random.randint(1, 101, size=20)

# Array 2 com 20 números inteiros aleatórios
array2 = np.random.randint(1, 101, size=20)

print("Array 1:", array1)
print("Array 2:", array2)

# Produto escalar
produto_escalar = np.dot(array1, array2)
print("Produto escalar:", produto_escalar)

# Norma (magnitude) de cada array
norma1 = np.linalg.norm(array1)
norma2 = np.linalg.norm(array2)

# Produto escalar normalizado (cosseno do ângulo)
cosseno = produto_escalar / (norma1 * norma2)
print("Cosseno do ângulo:", cosseno)

# Ângulo em radianos
angulo_rad = np.arccos(cosseno)
print("Ângulo (radianos):", angulo_rad)

# Ângulo em graus
angulo_graus = np.degrees(angulo_rad)
print("Ângulo (graus):", angulo_graus)
```

### Exercício 3
Encontre as raízes da função f(x) = x⁴ - 3x³ + 2 utilizando SciPy.

```python
# Defino a função f(x) = x⁴ - 3x³ + 2.
# Uso root do scipy.optimize para encontrar as raízes (valores onde f(x) = 0).
# root é mais moderno e robusto que fsolve, converge melhor e não gera warnings. Testei com fsolve e me retornou erros no output
# scipy.optimize tem funções para otimização e resolução de equações.
# Testei com valores iniciais diferentes para encontrar raízes diferentes.

from scipy.optimize import root

def f(x):
    return x**4 - 3*x**3 + 2

raiz1 = root(f, 0)
raiz2 = root(f, 1)
raiz3 = root(f, 2)
raiz4 = root(f, 3)

print("Raiz 1:", raiz1.x)
print("Raiz 2:", raiz2.x)
print("Raiz 3:", raiz3.x)
print("Raiz 4:", raiz4.x)
```

### Exercício 4
Use a função de otimização de SciPy para maximizar f(x) = -x² + 4x + 1.

```python
# Defino a função f(x) = -x² + 4x + 1.
# Para maximizar, preciso minimizar a função negativa: -f(x).
# Uso minimize do scipy.optimize para encontrar o máximo.
# minimize encontra o valor de x que minimiza a função.

from scipy.optimize import minimize

def f(x):
    return -((-x**2 + 4*x + 1))

resultado = minimize(f, 0)

print("Valor de x que maximiza:", resultado.x)
print("Valor máximo:", -resultado.fun)
```

### Exercício 10
Calcule a integral dupla da função f(x,y) = xy + x² no intervalo [0,2] para x e [0,1] para y.

```python
# Defino a função f(x,y) = xy + x².
# Uso integrate.dblquad do scipy para calcular a integral dupla.
# dblquad integra em relação a y primeiro, depois em relação a x.
# Intervalos: x de 0 a 2, y de 0 a 1.
# dblquad retorna (resultado, erro), uso _ para ignorar o erro. --> caso seja necessário a resposta do estimado é só tirar "_"

from scipy import integrate

# Defino a função f(x,y)
def f(y, x):
    return x*y + x**2

# Integral dupla
resultado, _ = integrate.dblquad(f, 0, 2, 0, 1)

print("Integral dupla:", resultado)
```

### Exercício 16
Resolva a equação diferencial de segunda ordem d²y/dx² + y = 0, com condições iniciais y(0)=1 e y'(0)=0.

```python
# Defino a equação diferencial de segunda ordem: d²y/dx² + y = 0.
# Converto em sistema de equações de primeira ordem para usar odeint.
# Uso np.linspace para criar um intervalo divido em 100 partes iguais
# Uso odeint do scipy para resolver.
# Condições iniciais: y(0)=1 e y'(0)=0.


from scipy.integrate import odeint
import numpy as np

def sistema(y, x):
    y1, y2 = y
    dy1dx = y2
    dy2dx = -y1
    return [dy1dx, dy2dx]

y0 = [1, 0]
x = np.linspace(0, 10, 100)
y = odeint(sistema, y0, x)

print("Solução:")
print(y[:5]) #printa os primeiros 5 elementos do array
```

### Exercício 8
Plote um gráfico de linha com intervalos de confiança (sombramento) para a função y = sin(x).

```python
# Ploto um gráfico de sin(x) com intervalo de confiança (sombramento).
# Uso fill_between() para criar a região sombreada.
# A região sombreada representa a margem de erro/confiança.

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

# Intervalo de confiança (exemplo: ±0.2)
erro = 0.2

plt.plot(x, y, label='sin(x)', color='blue')
plt.fill_between(x, y - erro, y + erro, alpha=0.3, color='blue', label='Intervalo de confiança')
plt.xlabel('x')
plt.ylabel('y')
plt.title('sin(x) com intervalo de confiança')
plt.legend()
plt.show()
```
